## 1. 硬件进化：15 年 36 万倍

比特币挖矿硬件经历了人类科技史上最快的一次军备竞赛：

| 年代 | 硬件 | 算力 | 功耗 | 能效 (J/GH) |
|------|------|------|------|-------------|
| 2009 | CPU (Core 2) | ~10 MH/s | 65W | 6,500 |
| 2010 | GPU (Radeon 5870) | ~400 MH/s | 200W | 500 |
| 2012 | FPGA | ~1 GH/s | 80W | 80 |
| 2013 | ASIC 28nm (S1) | 180 GH/s | 360W | 2 |
| 2016 | ASIC 16nm (S9) | 14 TH/s | 1,372W | 0.1 |
| 2020 | ASIC 7nm (S19) | 110 TH/s | 3,250W | 0.03 |
| 2024 | ASIC 5nm (S21) | 200 TH/s | 3,500W | 0.018 |

**从 6,500 J/GH 到 0.018 J/GH，能效提升了 36 万倍。**

这是任何技术领域都罕见的指数级进步。摩尔定律（每 2 年翻倍）在这个速度面前
显得很慢。

````python
import matplotlib.pyplot as plt
import numpy as np

# 硬件数据
hardware = [
    ("CPU (2009)",    0.00001, 65,   "blue"),
    ("GPU (2010)",    0.0004,  200,  "green"),
    ("FPGA (2012)",   0.001,   80,   "orange"),
    ("S1 (2013)",     0.18,    360,  "red"),
    ("S9 (2016)",     14,      1372, "purple"),
    ("S19 (2020)",    110,     3250, "brown"),
    ("S21 (2024)",    200,     3500, "black"),
]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

names = [h[0] for h in hardware]
th_s   = [h[1] for h in hardware]  # TH/s
colors = [h[3] for h in hardware]
j_per_gh = [h[2]/h[1]*1000 for h in hardware]  # W/TH → J/GH

ax1.barh(names, th_s, color=colors, edgecolor="black")
ax1.set_xscale("log")
ax1.set_xlabel("算力 (TH/s, 对数轴)")
ax1.set_title("挖矿硬件进化：算力")

ax2.barh(names, j_per_gh, color=colors, edgecolor="black")
ax2.set_xscale("log")
ax2.set_xlabel("能效 (J/GH, 对数轴, 越低越好)")
ax2.set_title("挖矿硬件进化：能效")

plt.tight_layout()
plt.savefig("../data/hardware_evolution.png", dpi=100, bbox_inches="tight")
plt.close("all")
print("✅ 图表已保存到 data/hardware_evolution.png")
print()
print("💡 对数轴：每一条线 = 10 倍差距")
print("💡 15 年 36 万倍，这速度超越了摩尔定律")  
````

## 3. 能源争议

比特币挖矿年耗电约 **150 TWh**（≈ 阿根廷全国用电）。

**批评方认为**：
- 浪费电力用于无意义的数学计算
- 一笔 BTC 交易耗电 ≈ 数十万笔 Visa 交易
- 碳排放加剧气候变化

**辩护方认为**：
- 大量使用弃电（水电过剩期、天然气伴生气、太阳能富余）
- 比特币矿场跟着最便宜的电跑 → 天然倾向可再生能源
- 银行系统、金矿开采、印钞也消耗大量能源
- 换来的是一条去中心化、无需信任的全球结算网络

**诚实说**：两边都能挑出支持自己观点的数据。这不是一个非黑即白的问题。

个人看法：如果你认为比特币提供的"无需许可的全球价值传输"有价值，
那它的能耗是有意义的。如果你不认为——那它确实浪费。

## 4. 减半与未来

比特币总量上限 2100 万枚。区块奖励每 21 万个块减半（≈ 4 年）：

```
2024 年：3.125 BTC/块  (年通胀 ~0.8%)
2028 年：1.5625 BTC/块  (年通胀 ~0.4%)
2032 年：0.78125 BTC/块 (年通胀 ~0.2%)
...
2140 年：最后一聪被挖出
```

**关键问题**：当区块奖励趋近于零，矿工的收入将主要来自**交易手续费**。

比特币安全模型的大前提是：
> 手续费收入能维持足够的全网算力，使 51% 攻击在经济上不可行。

这仍然是一个**开放问题**。可能的发展方向：
- 比特币成为结算层，大量交易走 Layer 2（闪电网络），主链手续费高涨
- 矿工收入结构转型，手续费占比上升到 50%+
- 如果手续费不够，安全预算可能下降

没有人知道 100 年后比特币会怎样——中本聪也不知道。
