## 8.1 硬件进化：从 CPU 到 ASIC

15 年间能效提升了 **360,000 倍**。这是科技史上前所未有的速度。

| 时代 | 硬件 | 算力 | 功耗 | 能效 (J/GH) |
|------|------|------|------|-------------|
| 2009 | CPU | ~10 MH/s | 65W | 6500 |
| 2010 | GPU 5870 | ~400 MH/s | 200W | 500 |
| 2012 | FPGA | ~1 GH/s | 80W | 80 |
| 2013 | ASIC 28nm | 180 GH/s | 360W | 2 |
| 2016 | ASIC 16nm | 14 TH/s | 1372W | 0.1 |
| 2020 | ASIC 7nm | 110 TH/s | 3250W | 0.03 |
| 2024 | ASIC 5nm | 200 TH/s | 3500W | 0.018 |

> 今天一台 S21 的算力 ≈ 2009 年全网算力的 2000 万倍。

````python
import matplotlib.pyplot as plt
import numpy as np

hardware = [
    ("CPU (2009)",    0.00001, 65, "blue"),
    ("GPU (2010)",    0.0004, 200, "green"),
    ("FPGA (2012)",   0.001, 80, "orange"),
    ("S1 (2013)",     0.18, 360, "red"),
    ("S9 (2016)",     14, 1372, "purple"),
    ("S19 (2020)",    110, 3250, "brown"),
    ("S21 (2024)",    200, 3500, "black"),
]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

names = [h[0] for h in hardware]
hashrates = [h[1] for h in hardware]  # TH/s
colors = [h[3] for h in hardware]
efficiency = [h[2]/h[1]*1000 for h in hardware]  # W/TH -> J/GH

ax1.barh(names, hashrates, color=colors, edgecolor="black")
ax1.set_xscale("log")
ax1.set_xlabel("Hashrate (TH/s, log scale)")
ax1.set_title("Mining Hardware Evolution: Hashrate")

ax2.barh(names, efficiency, color=colors, edgecolor="black")
ax2.set_xscale("log")
ax2.set_xlabel("Efficiency (J/GH, log scale)")
ax2.set_title("Mining Hardware Evolution: Efficiency")

plt.tight_layout()
plt.savefig("../data/hardware_evolution.png", dpi=100, bbox_inches="tight")
plt.savefig(f"../data/plot_{f.replace(".ipynb","")}.png", dpi=100, bbox_inches="tight")
plt.close("all")
print("✅ Saved to data/hardware_evolution.png")
````

## 8.3 能源争议

比特币挖矿年耗电 ~150 TWh（≈ 阿根廷全国用电）。

**批评方：**
- 浪费电力，碳排放
- 一笔 BTC 交易耗电 ≈ 70 万笔 Visa 交易

**辩护方：**
- 大量使用弃电（水电过剩、天然气伴生气）
- 比特币激励可再生能源投资（矿场跟着最便宜的电跑 → 开发弃电）
- 传统金融系统（银行、金矿、印钞）也耗能
- 换来的是一条无需信任的全球结算网络

**没有标准答案。** 两边都能找到支持自己观点的数据。
